```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## **Read in Memory Data**

Test 1 is run for 1 Myr with the memory leak (scipy > 1.16.3). Test 2 is run for 1 Myr with the memory leak fixed (scipy = 1.16.3). Test 3 is run for 3 Myr using the Balogh+2026 model.

```python
directory = "/Users/ben/Downloads/Planet Formation/DiscEvolution Simulations/Tests/"

test_number = 1

quixote_memory_log = pd.read_csv(f"{directory}quixote_test_{test_number}_memory_log.csv")
mac_memory_log = pd.read_csv(f"{directory}mac_test_{test_number}_memory_log.csv")

quixote_planets_log = pd.read_csv(f"{directory}quixote_test_{test_number}_planet_implant_log.csv")
mac_planets_log = pd.read_csv(f"{directory}mac_test_{test_number}_planet_implant_log.csv")
```

```python
n = quixote_memory_log['n [steps]']

quixote_mem_current = quixote_memory_log['mem_current [MB]']
mac_mem_current = mac_memory_log['mem_current [MB]']

quixote_mem_peak = quixote_memory_log['mem_peak [MB]']
mac_mem_peak = mac_memory_log['mem_peak [MB]']

quixote_n_planets = quixote_planets_log['n [steps]']
mac_n_planets = mac_planets_log['n [steps]']

quixote_delta_t_clock = quixote_memory_log['delta_t_clock [ms]']
mac_delta_t_clock = mac_memory_log['delta_t_clock [ms]']

quixote_eta = quixote_memory_log['time_remaining [hr]']
mac_eta = mac_memory_log['time_remaining [hr]']

quixote_elapsed_time = quixote_memory_log['elapsed_clock [hr]']
mac_elapsed_time = mac_memory_log['elapsed_clock [hr]']
```

## **Plot Current Memory Usage**

```python
fig, ax = plt.subplots(figsize=(8, 6))

ax.plot(n, quixote_mem_current, label = 'quixote', color = 'blue')
ax.plot(n, mac_mem_current, label = 'Mac', color = 'red')

for i in range(len(quixote_n_planets)):
    ax.axvline(x = quixote_n_planets[i], color = 'black', linestyle = '--', alpha = 0.5)

ax.set_xlabel('Number of Steps')
ax.set_ylabel('Current Memory Usage [MB]')
ax.legend()

plt.show()
```

## **Plot Peak Memory Usage**

```python
fig, ax = plt.subplots(figsize=(8, 6))

ax.plot(n, quixote_mem_peak, label = 'quixote', color = 'blue')
ax.plot(n, mac_mem_peak, label = 'Mac', color = 'red')

for i in range(len(quixote_n_planets)):
    ax.axvline(x = quixote_n_planets[i], color = 'black', linestyle = '--', alpha = 0.5)

ax.set_xlabel('Number of Steps')
ax.set_ylabel('Peak Memory Usage [MB]')
ax.legend()

plt.show()
```

## **Plot Clock Time per Step**

```python
fig, ax = plt.subplots(figsize=(8, 6))

ax.plot(n, quixote_delta_t_clock, label = 'quixote', color = 'blue')
ax.plot(n, mac_delta_t_clock, label = 'Mac', color = 'red')

ax.set_xlabel('Number of Steps')
ax.set_ylabel('Clock Time per Step [ms]')
ax.legend()

plt.show()
```

## **Plot Estimated Time Remaining**

```python
fig, ax = plt.subplots(figsize=(8, 6))

ax.plot(n, quixote_eta, label = 'quixote', color = 'blue')
ax.plot(n, mac_eta, label = 'Mac', color = 'red')

ax.set_xlabel('Number of Steps')
ax.set_ylabel('Time Remaining [hr]')
ax.legend()

plt.show()
```

## **Plot Elapsed Clock Time**

```python
fig, ax = plt.subplots(figsize=(8, 6))

ax.plot(n, quixote_elapsed_time, label = 'quixote', color = 'blue')
ax.plot(n, mac_elapsed_time, label = 'Mac', color = 'red')

ax.set_xlabel('Number of Steps')
ax.set_ylabel('Elapsed Time [hr]')
ax.legend()

plt.show()
```
